# Блок метрик: Оценка качества моделей

## Вопрос 1: Как рассчитывается F1-мера (F1-score) и почему она предпочтительнее простого среднего арифметического между Precision и Recall?

**Ответ:** F1-мера — это гармоническое среднее между точностью (Precision) и полнотой (Recall). Обычное среднее арифметическое маскировало бы плохую работу модели, если одна из метрик близка к нулю (например, модель не должна получать хорошую оценку, если у нее идеальный Recall, но нулевой Precision). Гармоническое среднее сильно штрафует за дисбаланс между этими метриками.

### Формулы:

Сначала определяются базовые метрики:

$$Precision = \frac{TP}{TP + FP}$$

$$Recall = \frac{TP}{TP + FN}$$

Итоговая формула F1-меры:

$$F1 = 2 \cdot \frac{Precision \cdot Recall}{Precision + Recall}$$

---

## Вопрос 2: В чем разница между MAE, MSE и RMSE в задачах регрессии? Как они реагируют на выбросы (outliers)?

**Ответ:** Все три метрики измеряют ошибку предсказания непрерывной величины (где $y_i$ — реальное значение, а $\hat{y}_i$ — предсказанное), но делают это с разной чувствительностью к аномалиям.

**MAE (Mean Absolute Error):** Средняя абсолютная ошибка. Она растет линейно и менее чувствительна к выбросам, так как берет просто модуль отклонения.

$$MAE = \frac{1}{n} \sum_{i=1}^{n} |y_i - \hat{y}_i|$$

**MSE (Mean Squared Error):** Среднеквадратичная ошибка. Сильно штрафует модель за большие ошибки (выбросы), так как возводит отклонение в квадрат.

$$MSE = \frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2$$

**RMSE (Root Mean Squared Error):** Корень из MSE. Имеет ту же размерность, что и целевая переменная (например, рубли или метры), что делает ее удобной для интерпретации бизнесом, но она сохраняет высокую чувствительность к выбросам.

$$RMSE = \sqrt{\frac{1}{n} \sum_{i=1}^{n} (y_i - \hat{y}_i)^2}$$

---

## Вопрос 3: Из каких показателей строится ROC-кривая и как они вычисляются?

**Ответ:** ROC-кривая показывает зависимость между долей истинно положительных и ложноположительных срабатываний при изменении порога классификации (threshold). Она строится в осях FPR (ось X) и TPR (ось Y).

**TPR (True Positive Rate)** — это синоним Recall. Показывает долю правильно найденных положительных объектов среди всех реальных положительных:

$$TPR = \frac{TP}{TP + FN}$$

**FPR (False Positive Rate)** — показывает долю ложных срабатываний среди всех реальных отрицательных объектов:

$$FPR = \frac{FP}{FP + TN}$$

---

## Вопрос 4: Зачем используется Log Loss (бинарная кросс-энтропия), если есть обычная метрика Accuracy?

**Ответ:** Accuracy просто считает долю правильных ответов (жесткая классификация 0 или 1), тогда как Log Loss оценивает уверенность модели в своих предсказаниях (вероятность от 0 до 1). Если модель угадывает класс, но сомневается (например, выдает вероятность $0.51$ для класса $1$), Log Loss оштрафует её сильнее, чем если бы она была уверена (выдала $0.99$).

### Формула:

(где $y_i$ — истинный класс $0$ или $1$, а $\hat{y}_i$ — предсказанная вероятность принадлежности к классу $1$):

$$LogLoss = - \frac{1}{N} \sum_{i=1}^{N} [y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i)]$$

---

## Вопрос 5: Как по матрице ошибок (Confusion Matrix) рассчитываются Precision (Точность) и Recall (Полнота), и что они физически означают?

**Ответ:** Обе метрики строятся на основе четырех исходов: $TP$ (истинно положительные), $TN$ (истинно отрицательные), $FP$ (ложноположительные) и $FN$ (ложноотрицательные).

**Precision** показывает степень доверия к модели, когда она говорит «Да». Из всех объектов, которые алгоритм отнес к положительному классу, какая доля действительно оказалась положительной?

**Recall** показывает способность алгоритма находить нужный класс вообще. Из всех реально положительных объектов в данных, какую долю алгоритм смог обнаружить?

### Формулы:

$$Precision = \frac{TP}{TP + FP}$$

$$Recall = \frac{TP}{TP + FN}$$

---

## Вопрос 6: Что такое mAP (Mean Average Precision) и как она вычисляется?

**Ответ:** mAP — это усредненная средняя точность. Это комплексная метрика, которая оценивает качество алгоритмов, где важен порядок выдачи (рекомендательные системы, поисковики) или где нужно находить несколько объектов разных классов на одном изображении (Object Detection).

Чтобы понять mAP, нужно сначала понять AP (Average Precision). AP — это, по сути, площадь под кривой Precision-Recall для одного конкретного класса или запроса. Она показывает, насколько хорошо модель сохраняет высокую точность (Precision) по мере увеличения полноты (Recall).

Когда мы посчитали AP для каждого отдельного класса (например, для кошек, собак и машин в задаче распознавания) или для каждого поискового запроса пользователя, мы просто усредняем эти значения. Это и есть mAP.

### Формулы:

Сначала считается Average Precision ($AP$) для одного класса. В дискретном виде это сумма точностей при каждом изменении полноты:

$$AP = \sum_{k=1}^{n} (Recall_k - Recall_{k-1}) \cdot Precision_k$$

Затем вычисляется mAP как среднее арифметическое всех $AP$ по всем $N$ классам (или запросам):

$$mAP = \frac{1}{N} \sum_{i=1}^{N} AP_i$$

где $N$ — общее количество классов (в компьютерном зрении) или пользователей/запросов (в рекомендательных системах).

# Блок методов: Линейная и логистическая регрессия

## Вопрос 1: Как математически формулируется предсказание множественной линейной регрессии и какую функцию потерь она минимизирует при классическом обучении?

**Ответ:** Модель предсказывает целевую непрерывную переменную как линейную комбинацию признаков, умноженных на их веса, плюс свободный член (bias). Для обучения алгоритм чаще всего использует метод наименьших квадратов (Ordinary Least Squares, OLS), минимизируя среднеквадратичную ошибку (MSE).

### Формула предсказания:

$$\hat{y} = w_0 + w_1 x_1 + w_2 x_2 + \dots + w_n x_n = w^T x$$

где $w$ — вектор весов, $x$ — вектор признаков объекта (включая константу $x_0 = 1$ для $w_0$).

### Функция потерь (MSE):

$$L(w) = \frac{1}{N} \sum_{i=1}^{N} (y_i - w^T x_i)^2$$

---

## Вопрос 2: В чем математическая разница между Ridge (L2) и Lasso (L1) регуляризацией в линейной регрессии, и как они по-разному влияют на веса модели?

**Ответ:** Регуляризация нужна для борьбы с переобучением (особенно при мультиколлинеарности, когда признаки сильно коррелируют). Она добавляет в функцию потерь «штраф» за слишком большие веса.

**Ridge (L2)** штрафует за сумму квадратов весов. Это приводит к тому, что веса уменьшаются (стягиваются к нулю), но редко становятся точно равными нулю.

**Lasso (L1)** штрафует за сумму модулей весов. Ее главная особенность математически заключается в том, что она может строго занулять веса неинформативных признаков, работая как встроенный механизм отбора признаков (feature selection).

### Формула Ridge (L2):

$$L_{Ridge}(w) = MSE + \lambda \sum_{j=1}^{n} w_j^2$$

### Формула Lasso (L1):

$$L_{Lasso}(w) = MSE + \lambda \sum_{j=1}^{n} |w_j|$$

(где $\lambda$ — гиперпараметр, контролирующий силу регуляризации).

---

## Вопрос 3: Почему в логистической регрессии для задачи классификации используется сигмоида (логистическая функция), а не просто линейное уравнение?

**Ответ:** Линейная регрессия выдает значения в диапазоне от $-\infty$ до $+\infty$. Это не имеет физического смысла, если нам нужно предсказать вероятность принадлежности к классу (которая обязана лежать в пределах от $0$ до $1$). Сигмоида элегантно решает эту проблему: она принимает на вход любое вещественное число (результат линейной комбинации $w^T x$) и нелинейно "сжимает" его в интервал $(0, 1)$.

### Формула сигмоиды:

$$\sigma(z) = \frac{1}{1 + e^{-z}}$$

### Предсказание вероятности в логистической регрессии:

$$P(y=1|x) = \sigma(w^T x) = \frac{1}{1 + e^{-w^T x}}$$

---

## Вопрос 4: Почему логистическая регрессия оптимизирует Log Loss (бинарную кросс-энтропию), а не MSE?

**Ответ:** Если подставить сигмоиду $\sigma(w^T x)$ напрямую в формулу MSE, итоговая функция потерь станет невыпуклой (non-convex). В ней появится множество локальных минимумов, и алгоритм градиентного спуска с большой вероятностью застрянет, не найдя оптимальные веса. Функция Log Loss выведена из метода максимального правдоподобия (Maximum Likelihood Estimation) — она является строго выпуклой (один глобальный минимум), что гарантирует успешную сходимость градиентного спуска.

### Формула Log Loss:

$$L(w) = - \frac{1}{N} \sum_{i=1}^{N} \left[ y_i \log(\hat{y}_i) + (1 - y_i) \log(1 - \hat{y}_i) \right]$$

(где $\hat{y}_i = \sigma(w^T x_i)$)

---

## Вопрос 5: Как градиентный спуск обновляет веса $w$ на каждом шаге обучения?

**Ответ:** Градиентный спуск итеративно корректирует веса модели в направлении, противоположном градиенту функции потерь (так как градиент указывает направление наискорейшего роста функции, а нам нужно ее минимизировать). Размер шага определяется гиперпараметром $\eta$ (learning rate, скорость обучения).

### Формула обновления весов:

$$w_{new} = w_{old} - \eta \nabla L(w_{old})$$

(где $\nabla L(w_{old})$ — вектор частных производных функции потерь по каждому весу).

# Блок алгоритмов: Деревья решений (CART)

## 1. Общий принцип работы (Жадный алгоритм)

CART работает по принципу рекурсивного разбиения пространства признаков. Он использует так называемый «жадный» (greedy) подход:

1. Алгоритм берет весь датасет (корневой узел).
2. Перебирает все доступные признаки и все возможные пороги (значения) для каждого признака.
3. Для каждого варианта разбиения он вычисляет, насколько «чище» или «однороднее» станут два получившихся дочерних узла по сравнению с родительским.
4. Выбирает тот признак и тот порог, который дает максимальный прирост информации (максимально снижает ошибку).
5. Повторяет этот процесс рекурсивно для каждого нового узла, пока не сработает критерий остановки (например, достигнута максимальная глубина дерева max_depth или в листе осталось слишком мало объектов min_samples_leaf).

---

## 2. Математика разбиения: Функция потерь

Чтобы алгоритм мог понять, какое разбиение «лучше», ему нужна метрика (функция потерь). В терминах деревьев это называется критерием информативности (Impurity Function).

Общая формула качества разбиения узла $Q$ на два подузла $Q_{left}$ и $Q_{right}$ выглядит так:

$$Gain = H(Q) - \left( \frac{|Q_{left}|}{|Q|} H(Q_{left}) + \frac{|Q_{right}|}{|Q|} H(Q_{right}) \right)$$

Где:
- $H(Q)$ — значение критерия информативности (ошибки) в родительском узле.
- $|Q|$ — количество объектов в узле.

Алгоритм ищет такое разбиение, при котором значение $Gain$ будет максимальным (то есть мы вычитаем как можно меньшую взвешенную ошибку дочерних узлов).

А вот сама функция $H(Q)$ зависит от того, какую задачу мы решаем.

---

## 3. CART для задачи Классификации

В классификации мы хотим, чтобы в итоговых листьях оказались объекты одного класса (чтобы узел был «чистым»). CART по умолчанию использует Критерий Джини (Gini Impurity).

### Критерий Джини

Gini показывает вероятность того, что случайно взятый из узла объект будет неверно классифицирован, если мы будем предсказывать класс случайно, опираясь на распределение классов в этом узле.

$$H_{Gini}(Q) = 1 - \sum_{k=1}^{K} p_k^2$$

Где:
- $K$ — количество классов.
- $p_k$ — доля объектов класса $k$ в узле $Q$.

### Как это работает:

- Если в узле все объекты принадлежат к одному классу (идеальная чистота), то $p_1 = 1$, и $H_{Gini} = 1 - 1^2 = 0$. Ошибка равна нулю!
- Если в узле объекты двух классов распределены 50/50, то $p_1 = 0.5, p_2 = 0.5$, и $H_{Gini} = 1 - (0.5^2 + 0.5^2) = 0.5$. Это худший вариант (максимальная неопределенность).

> **Примечание:** Иногда на собеседованиях спрашивают про Энтропию Шеннона. CART может использовать и ее, но исторически Энтропия чаще ассоциируется с алгоритмами ID3/C4.5. Ее формула: $H_{Entropy}(Q) = - \sum_{k=1}^{K} p_k \log_2(p_k)$.

---

## 4. CART для задачи Регрессии

В регрессии целевая переменная непрерывна. Мы не можем считать «доли классов». Вместо этого предсказанием в листе (конечным ответом) будет просто среднее значение целевой переменной всех объектов, попавших в этот лист.

Поэтому критерием информативности $H(Q)$ здесь выступает дисперсия (Variance) или, что эквивалентно, MSE (Mean Squared Error) внутри узла.

### MSE для узла

Алгоритм пытается разбить данные так, чтобы в каждом дочернем узле значения целевой переменной были максимально похожи друг на друга (минимальный разброс вокруг среднего).

$$H_{MSE}(Q) = \frac{1}{|Q|} \sum_{i \in Q} (y_i - \bar{y}_Q)^2$$

Где:
- $y_i$ — истинное значение объекта $i$.
- $\bar{y}_Q$ — среднее значение ответов для всех объектов в узле $Q$ (это и есть предсказание модели для данного узла).

CART ищет такое разбиение, чтобы сумма дисперсий в левом и правом узлах была минимальной.

# Блок алгоритмов: Ансамблевые методы

## Введение

Одно дерево решений (CART) легко переобучается и часто работает нестабильно. Чтобы это исправить, в машинном обучении используют ансамбли — объединение множества деревьев.

Случайный лес и Градиентный бустинг — это два главных ансамблевых метода, но их философия прямо противоположна.

---

## 1. Случайный лес (Random Forest)

### Идея: "Мудрость толпы"

Если собрать много независимых, но не очень точных экспертов (деревьев) и усреднить их ответы, общий результат будет точным и стабильным.

### Как это работает (Бэггинг + Метод случайных подпространств)

**Бэггинг (Bootstrap Aggregating):** Алгоритм создает множество глубоких деревьев (сотни или тысячи). Каждое дерево обучается независимо и параллельно на случайной подвыборке данных (с возвращением).

**Случайные подпространства:** При каждом разбиении узла внутри дерева алгоритм смотрит не на все признаки, а только на их случайную часть (обычно $\sqrt{N}$ для классификации). Это заставляет деревья быть максимально непохожими друг на друга (декоррелированными).

**Голосование:** Итоговый ответ получается путем усреднения ответов всех деревьев (для регрессии) или голосованием большинства (для классификации).

### Смысл

Каждое глубокое дерево сильно переобучено, но из-за того, что они разные, при усреднении их ошибки взаимно уничтожаются. Лес снижает дисперсию (variance) модели.

---

## 2. Градиентный бустинг (Gradient Boosting)

### Идея: "Работа над ошибками"

Вместо того чтобы строить деревья независимо, давайте строить их последовательно, так, чтобы каждое новое дерево исправляло косяки предыдущих.

### Как это работает

1. Сначала строится очень простое базовое предсказание (например, среднее значение целевой переменной по всему датасету).

2. Вычисляются ошибки (в математике бустинга — антиградиенты или псевдоостатки) — разница между реальными ответами и текущими предсказаниями ансамбля.

3. Строится новое, неглубокое дерево (weak learner, обычно глубиной 3-7), но оно предсказывает не саму целевую переменную, а ошибки предыдущего шага.

4. Ответ нового дерева умножается на скорость обучения (learning rate, $\eta$) и добавляется к общему предсказанию.

5. Процесс повторяется сотни раз.

### Математика

Шаг обновления ансамбля $F$ на итерации $m$ выглядит так:

$$F_m(x) = F_{m-1}(x) + \eta \cdot h_m(x)$$

Где $h_m(x)$ — новое дерево, обученное на остатках, а $\eta$ — learning rate (обычно малое число, например, 0.1).

### Смысл

Бустинг шаг за шагом "дотягивает" предсказания до идеала. Он снижает смещение (bias) модели.

---

## 3. Главные отличия

| Характеристика | Случайный лес (Random Forest) | Градиентный бустинг (Gradient Boosting) |
|---|---|---|
| Построение деревьев | Параллельное и независимое. | Последовательное (каждое следующее зависит от предыдущих). |
| Глубина деревьев | Глубокие (часто без ограничений), переобученные. | Мелкие (weak learners, глубина 3-7), недообученные. |
| Чувствительность к выбросам | Очень устойчив (выбросы теряются при усреднении). | Чувствителен (если не настроить правильную функцию потерь), так как алгоритм будет отчаянно пытаться исправить большую ошибку на выбросе. |
| Настройка гиперпараметров | Работает "из коробки". Главное — побольше деревьев. | Требует тщательной настройки (learning_rate, max_depth, n_estimators). Легко переобучить, если сделать слишком много шагов. |
| Точность | Хорошая, но часто уступает бустингу. | При правильной настройке — state-of-the-art алгоритм для табличных данных. |

## 1. XGBoost (eXtreme Gradient Boosting)

Исторически первый прорывной алгоритм (вышел в 2014 году), который задал стандарт индустрии.

### Главные фишки

**Математика:** В отличие от базового бустинга, который использует только первый градиент (скорость изменения ошибки), XGBoost раскладывает функцию потерь в ряд Тейлора до второго порядка (использует Гессиан — ускорение изменения ошибки). Это позволяет алгоритму сходиться к оптимуму быстрее и точнее.

**Рост деревьев:** Строит деревья в ширину (level-wise). Он сначала разбивает все узлы на одном уровне абстракции, и только потом переходит к следующему. Это работает стабильно, но может строить лишние, неинформативные ветки.

**Регуляризация:** Имеет встроенную L1 и L2 регуляризацию, что жестко контролирует переобучение.

### Пример кода

```python
import xgboost as xgb
from sklearn.metrics import accuracy_score

# XGBoost любит работать со своим оптимизированным форматом данных DMatrix
dtrain = xgb.DMatrix(X_train, label=y_train)
dtest = xgb.DMatrix(X_test, label=y_test)

params = {
    'objective': 'binary:logistic', # задача бинарной классификации
    'max_depth': 4,                 # максимальная глубина дерева
    'learning_rate': 0.1,           # скорость обучения
    'eval_metric': 'logloss'        # метрика оценки
}

# Обучение
xgb_model = xgb.train(params, dtrain, num_boost_round=100)

# Предсказание (выдает вероятности)
xgb_preds_proba = xgb_model.predict(dtest)
xgb_preds = (xgb_preds_proba > 0.5).astype(int)
print(f"XGBoost Accuracy: {accuracy_score(y_test, xgb_preds):.4f}")
```

## 2. LightGBM (Light Gradient Boosting Machine)

Разработка от Microsoft (2017 год). Создавалась как более быстрая и легковесная альтернатива XGBoost для работы с гигантскими датасетами.

### Главные фишки

**Рост деревьев:** Строит деревья в глубину (leaf-wise). Алгоритм ищет лист с наибольшей ошибкой и разбивает только его, игнорируя остальные листья на этом уровне, если они уже достаточно «чистые». Это дает лучшую точность при меньшем количестве узлов, но алгоритм сильнее склонен к переобучению (нужно строго ограничивать max_depth или num_leaves).

**Оптимизация (GOSS и EFB):** Использует градиентную одностороннюю выборку (отбрасывает объекты, на которых модель уже дает хороший прогноз, и учится только на сложных) и эксклюзивное объединение признаков (склеивает разреженные признаки, например, нули, чтобы не тратить на них память).

### Пример кода

```python
import lightgbm as lgb
from sklearn.metrics import accuracy_score

# Свой формат данных Dataset
train_data = lgb.Dataset(X_train, label=y_train)

params = {
    'objective': 'binary',
    'metric': 'binary_logloss',
    'learning_rate': 0.1,
    'num_leaves': 31, # Важнейший параметр для LightGBM вместо max_depth
}

# Обучение
lgb_model = lgb.train(params, train_data, num_boost_round=100)

# Предсказание
lgb_preds_proba = lgb_model.predict(X_test)
lgb_preds = (lgb_preds_proba > 0.5).astype(int)
print(f"LightGBM Accuracy: {accuracy_score(y_test, lgb_preds):.4f}")
```

## 3. CatBoost (Categorical Boosting)

Разработка от Яндекса (2017 год). Король табличных данных, где много текстовых или категориальных признаков (профессия, город, цвет и т.д.).

### Главные фишки

**Симметричные деревья (Oblivious Trees):** На одном уровне дерева для всех узлов используется один и тот же признак и один и тот же порог разбиения. Это делает модель невероятно быстрой при инференсе (предсказании на новых данных) и служит отличной защитой от переобучения.

**Обработка категорий из коробки:** Вам не нужно делать OHE (One-Hot Encoding) до обучения. Вы просто передаете названия колонок в модель. CatBoost использует Target Encoding, но с хитрым алгоритмом перемешивания (Ordered Target Encoding), чтобы избежать утечки данных (data leakage).

**Ordered Boosting:** Специальный механизм расчета градиентов, который предотвращает переобучение на ранних этапах.

### Пример кода

```python
from catboost import CatBoostClassifier
from sklearn.metrics import accuracy_score

# Указываем индексы или имена категориальных фичей
cat_features = [0, 2, 5] # пример индексов колонок с категориями

# CatBoost использует привычный sklearn-подобный API
cb_model = CatBoostClassifier(
    iterations=100,
    learning_rate=0.1,
    depth=6,
    loss_function='Logloss',
    cat_features=cat_features, # магия происходит здесь
    verbose=0 # отключаем лишний вывод текста при обучении
)

# Обучение
cb_model.fit(X_train, y_train)

# Предсказание (сразу выдает метки классов)
cb_preds = cb_model.predict(X_test)
print(f"CatBoost Accuracy: {accuracy_score(y_test, cb_preds):.4f}")
```

## Блок: Подготовка данных и Feature Engineering

### Вопрос 1: В чем разница между One-Hot Encoding (OHE) и Label Encoding? Когда и какой метод использовать?

**Ответ:** Это два основных способа перевести текстовые категории в числа, которые понимает модель.

**Label Encoding** присваивает каждой уникальной категории случайное целое число (например, "Москва" = 0, "Казань" = 1, "Уфа" = 2). Этот метод подходит только для признаков, где есть естественный порядок (ординальные признаки, например: "низкий", "средний", "высокий"), либо для алгоритмов на основе деревьев, которые умеют делать разбиения по числам. Если скормить такие данные линейной модели, она решит, что "Уфа" больше "Москвы" в два раза, что является бредом.

**One-Hot Encoding** создает для каждой категории свой отдельный бинарный столбец (0 или 1). Это обязательный подход для линейных моделей (регрессий) и метрических алгоритмов (например, KNN), чтобы не создавать ложных математических зависимостей между независимыми категориями.

---

### Вопрос 2: Что такое мультиколлинеарность, почему она опасна для линейных моделей и как с ней бороться?

**Ответ:** Мультиколлинеарность — это ситуация, когда два или более признака сильно коррелируют друг с другом (например, площадь квартиры в квадратных метрах и площадь в квадратных футах).

**Опасность:** В линейной регрессии это приводит к математической нестабильности (матрица признаков становится близкой к вырожденной). Веса модели начинают «скакать» и могут принимать гигантские значения, модель теряет интерпретируемость и легко переобучается.

**Как бороться:**
1. Удалить один из сильно коррелирующих признаков
2. Использовать L1-регуляризацию (Lasso, которая сама занулит лишний признак)
3. Применить алгоритмы снижения размерности (например, PCA)

---

### Вопрос 3: Как правильно обрабатывать пропуски в данных (NaN)? Почему нельзя просто везде подставить среднее значение?

**Ответ:** Заполнение всех пропусков средним или медианой — это быстрый, но часто плохой подход, потому что он искажает исходное распределение признака и занижает его дисперсию (variance).

**Альтернативы:**

1. Если пропусков очень мало (менее 1-2%), строки можно просто удалить.
2. Заполнение константой (например, −999), чтобы древесные модели могли выделить эти пропуски в отдельную ветку логики.
3. Использование ML-моделей для предсказания пропусков (например, KNN Imputer алгоритмически находит похожие объекты и берет их значения).

## Блок: Валидация модели

### Вопрос 4: Что такое кросс-валидация (Cross-Validation) и в каких случаях обязательно использовать Stratified K-Fold вместо обычного K-Fold?

**Ответ:** При обычном разбиении данных (train/test) мы рискуем случайно получить «удачное» или «неудачное» разбиение. Кросс-валидация K-Fold разбивает обучающую выборку на K равных частей (фолдов). Модель обучается K раз: каждый раз один из фолдов выступает в роли валидационного, а остальные K−1 — в роли тренировочных. Это дает более надежную и усредненную оценку качества.

**Stratified K-Fold** становится строго обязательным при дисбалансе классов (например, 95% нулей и 5% единиц). Он математически гарантирует, что в каждом из K фолдов пропорция классов останется точно такой же, как в исходном датасете. Обычный K-Fold может случайно собрать все редкие единицы в один фолд, и модель вообще не научится их распознавать.

## Блок: Обучение без учителя (Unsupervised Learning)

### Вопрос 5: Как работает алгоритм K-Means и в чем его фундаментальные отличия от DBSCAN?

#### K-Means (К-средних)

**Идея:** Алгоритм ищет заранее заданное количество кластеров (K). Он итеративно минимизирует расстояния между точками и центроидами кластеров.

**Алгоритм:**

1. **Инициализация:** Случайным образом выбрать K точек из датасета в качестве начальных центроидов $\mu_1, \mu_2, \ldots, \mu_K$.

2. **Назначение (Assignment step):** Для каждой точки $x_i$ найти ближайший центроид и присвоить точку соответствующему кластеру $C_k$:
   $$C_k = \{ x_i : \|x_i - \mu_k\| \leq \|x_i - \mu_j\| \text{ для всех } j \neq k \}$$
   где $\|x_i - \mu_k\|$ — Евклидово расстояние.

3. **Обновление (Update step):** Пересчитать центроиды как центр масс каждого кластера:
   $$\mu_k^{new} = \frac{1}{|C_k|} \sum_{x_i \in C_k} x_i$$

4. **Сходимость:** Повторять шаги 2 и 3 до тех пор, пока центроиды не перестанут изменяться (или пока не достигнуто максимальное число итераций).

**Функция потерь (inertia или WCSS — Within-Cluster Sum of Squares):**

$$J = \sum_{k=1}^{K} \sum_{x_i \in C_k} \|x_i - \mu_k\|^2$$

**Недостатки:**
- Нужно заранее знать количество кластеров K
- Очень чувствителен к выбросам (они тянут центроиды на себя)
- Может найти только кластеры шарообразной формы (сферически симметричные)
- Результат зависит от инициализации (может застрять в локальном минимуме)

---

#### DBSCAN (Density-Based Spatial Clustering of Applications with Noise)

**Идея:** Алгоритм группирует точки, которые плотно расположены друг к другу. Он автоматически находит количество кластеров и может обнаруживать выбросы как точки в разреженных регионах.

**Определения:**

- **ε-соседство (eps-neighborhood):** Множество всех точек в радиусе ε от точки $x_i$:
  $$N_\varepsilon(x_i) = \{ x_j : d(x_i, x_j) \leq \varepsilon \}$$

- **Основная точка (Core Point):** Если в ε-соседстве точки находится не менее MinPts соседей, то она основная.

- **Граничная точка (Border Point):** Точка, которая не является основной, но находится в ε-соседстве основной точки.

- **Шум (Noise/Outlier):** Точка, которая не основная и не граничная.

**Алгоритм:**

1. **Инициализация:** Выбрать случайную непосещённую точку $x_i$ из датасета.

2. **Проверка:** Если точка основная (имеет ≥ MinPts соседей в радиусе ε):
   - Создать новый кластер
   - Добавить все её ε-соседей в кластер
   - Рекурсивно применить проверку к каждому новому соседу

3. **Метки:** Если точка граничная, она присваивается кластеру, но не расширяет его. Если шум, метка Noise.

4. **Повторение:** Перейти к следующей непосещённой точке.

**Функция плотности (не минимизируется, а просто проверяется):**

Одна из интерпретаций — минимизация расстояния до MinPts-соседей:

$$\text{Density}(x_i) = \frac{\text{MinPts}}{\text{Volume of } \varepsilon\text{-neighborhood}}$$

**Преимущества:**
- Не нужно заранее указывать количество кластеров
- Находит кластеры любой сложной формы (полумесяцы, спирали, неправильные фигуры)
- Автоматически обнаруживает выбросы (помечает их как шум)
- Не чувствителен к выбросам (они остаются шумом)

**Недостатки:**
- Нужно правильно подобрать ε и MinPts (это не тривиально)
- Плохо работает с кластерами различной плотности
- Требует вычисления расстояний между всеми парами точек (медленнее на больших датасетах)

**Визуальное сравнение:** K-Means находит сферические кластеры. DBSCAN находит кластеры произвольной формы и очищает выбросы.

### Вопрос 6: В чем суть метода главных компонент (PCA) и какую главную задачу он решает?

**Ответ:** PCA — это линейный алгоритм снижения размерности. С точки зрения математики он находит новые ортогональные оси (главные компоненты) в пространстве признаков, вдоль которых дисперсия (разброс) данных максимальна.

**Как это работает:**

1. Алгоритм вычисляет матрицу ковариации признаков.
2. Находит 
собственные значения и собственные векторы этой матрицы.
3. Собственные векторы (компоненты) упорядочиваются по убыванию соответствующих собственных значений (дисперсии).
4. Первая компонента забирает на себя наибольшую долю дисперсии, вторая — наибольшую из оставшейся и так далее.

**Преобразование:**

Исходные данные $X$ проецируются на k первых главных компонент:

$$X_{transformed} = X \cdot W_k$$

где $W_k$ — матрица из первых k собственных векторов.

**Применение:**

- Сжимает данные, оставляя 95% информации при снижении размерности со 100 признаков до 10
- Полностью убирает мультиколлинеарность (новые компоненты ортогональны и не коррелируют)
- Ускоряет обучение моделей
- Позволяет визуализировать многомерные данные в 2D или 3D

## Вопрос 7: Что такое проблема затухающего градиента (Vanishing Gradient) и как функция активации ReLU помогает ее решить по сравнению с Сигмоидой?

**Ответ:** При обучении глубоких сетей методом обратного распространения ошибки градиенты последовательно перемножаются от конца сети к ее началу.

**Проблема с Сигмоидой:**

Производная сигмоиды всегда лежит в диапазоне от $0$ до $0.25$:

$$\sigma'(x) = \sigma(x)(1 - \sigma(x)) \in [0, 0.25]$$

При перемножении множества таких чисел градиент экспоненциально стремится к нулю. В результате веса первых слоев почти не обновляются, и сеть перестает учиться.

**Решение с ReLU:**

ReLU (Rectified Linear Unit) решает эту проблему. Ее математическая формула:

$$f(x) = \max(0, x) = \begin{cases} x & \text{если } x > 0 \\ 0 & \text{если } x \leq 0 \end{cases}$$

Производная ReLU:

$$f'(x) = \begin{cases} 1 & \text{если } x > 0 \\ 0 & \text{если } x < 0 \end{cases}$$

Производная ReLU равна строго $1$ для всех $x > 0$ и $0$ для $x < 0$. Градиент (равный единице) проходит сквозь слои без затухания, позволяя обучать очень глубокие сети.

## Вопрос 8: Объясните суть алгоритма обратного распространения ошибки (Backpropagation) математически.

**Ответ:** Backpropagation — это способ эффективно вычислить градиент функции потерь по каждому весу в нейронной сети, чтобы затем обновить эти веса с помощью градиентного спуска. Он базируется на математическом правиле дифференцирования сложной функции (Chain Rule).

**Правило цепочки (Chain Rule):**

Если у нас есть вес $w$, результат сумматора $z$, выход активации $a$ и итоговая ошибка $L$, то влияние веса $w$ на ошибку $L$ вычисляется по цепочке с конца в начало:

$$\frac{\partial L}{\partial w} = \frac{\partial L}{\partial a} \cdot \frac{\partial a}{\partial z} \cdot \frac{\partial z}{\partial w}$$

**Процесс:**

1. Сначала вычисляется ошибка в последнем слое: $\frac{\partial L}{\partial a_{output}}$
2. Затем благодаря цепному правилу эта ошибка «распространяется» (backpropagates) назад через все слои
3. На каждом слое мы получаем $\frac{\partial L}{\partial w_i}$ для каждого веса $w_i$
4. Веса обновляются по правилу: $w_{new} = w_{old} - \eta \cdot \frac{\partial L}{\partial w}$

Это позволяет обучать многослойные сети благодаря тому, что мы эффективно переиспользуем уже вычисленные производные вместо повторного пересчета для каждого слоя.

## Вопрос 9: Зачем в архитектуры добавляют слои Batch Normalization (пакетной нормализации)?

**Ответ:** В процессе обучения распределение данных на входах скрытых слоев постоянно меняется из-за обновления весов предыдущих слоев (это называется внутренним ковариантным сдвигом). Batch Norm нормализует выходы предыдущего слоя (или входы следующего) в пределах одного мини-батша, приводя их к нулевому среднему и единичной дисперсии.

**Формула нормализации:**

Для значения $x_i$ в батче:

$$\hat{x}_i = \frac{x_i - \mu_B}{\sqrt{\sigma_B^2 + \epsilon}}$$

где:
- $\mu_B$ — среднее значение по батчу
- $\sigma_B^2$ — дисперсия батча
- $\epsilon$ — маленькое число (например, $10^{-5}$) для предотвращения деления на ноль

**После нормализации** алгоритм применяет обучаемые параметры сдвига $\gamma$ и масштаба $\beta$:

$$y_i = \gamma \hat{x}_i + \beta$$

**Преимущества:**

- Позволяет использовать более высокую скорость обучения (learning rate)
- Делает сеть менее чувствительной к начальной инициализации весов
- Работает как легкая регуляризация
- Ускоряет сходимость обучения

---

## Вопрос 10: В чем концептуальная разница между классическим SGD (Стохастическим градиентным спуском) и оптимизатором Adam?

**Ответ:** 

**SGD (Stochastic Gradient Descent):**

SGD обновляет веса, опираясь исключительно на градиент, вычисленный на текущем мини-батче:

$$w_{new} = w_{old} - \eta \cdot \nabla L(w_{old})$$

Он может сильно "блуждать" по оврагам функции потерь и медленно сходиться.

**Adam (Adaptive Moment Estimation):**

Adam — это продвинутый алгоритм, который вычисляет индивидуальную скорость обучения для каждого отдельного параметра. Он хранит экспоненциально скользящее среднее прошлых градиентов (момент первого порядка, аналог инерции) и квадратов прошлых градиентов (момент второго порядка).

**Формулы Adam:**

1. Момент первого порядка (скользящее среднее градиентов):
   $$m_t = \beta_1 m_{t-1} + (1 - \beta_1) \nabla L(w_t)$$

2. Момент второго порядка (скользящее среднее квадратов градиентов):
   $$v_t = \beta_2 v_{t-1} + (1 - \beta_2) (\nabla L(w_t))^2$$

3. Смещение (bias correction):
   $$\hat{m}_t = \frac{m_t}{1 - \beta_1^t}, \quad \hat{v}_t = \frac{v_t}{1 - \beta_2^t}$$

4. Обновление весов (адаптивная скорость обучения):
   $$w_{new} = w_{old} - \eta \cdot \frac{\hat{m}_t}{\sqrt{\hat{v}_t} + \epsilon}$$

где $\beta_1 \approx 0.9$, $\beta_2 \approx 0.999$ — коэффициенты экспоненциального сглаживания.

**Преимущества Adam:**

Это позволяет Adam автоматически адаптироваться к геометрии функции потерь:
- Делает большие шаги по плоским участкам
- Делает аккуратные шаги по крутым склонам

## Вопрос 11: Почему для обработки изображений используют Сверточные нейронные сети (CNN), а не обычные полносвязные (Multi-Layer Perceptron, MLP)?

**Ответ:** Полносвязная сеть для картинки размером 1000×1000 пикселей потребовала бы миллионы весов для каждого нейрона первого скрытого слоя. Это ведет к моментальному переобучению и нехватке памяти. CNN решают эту проблему за счет трех принципов:

### 1. Разделение параметров (Parameter Sharing)

Один и тот же фильтр (сверточное ядро) проходит по всей площади изображения. Если фильтр научился искать вертикальные линии в левом верхнем углу, он найдет их и в правом нижнем с теми же весами.

**Формула свертки:**

Для изображения с пикселями, обозначаемого как матрица $I$, и ядра $K$ размером $k \times k$, результат свертки на позиции $(i, j)$:

$$O_{i,j} = \sum_{u=0}^{k-1} \sum_{v=0}^{k-1} I_{i+u, j+v} \cdot K_{u,v}$$

Количество параметров уменьшается с миллионов до нескольких сотен или тысяч.

### 2. Локальные связи (Local Connectivity)

Нейрон в CNN смотрит не на всю картинку сразу, а только на небольшую область (рецептивное поле, receptive field), что соответствует устройству зрительной коры животных.

### 3. Эквивариантность к сдвигу (Translational Equivariance)

Смещение объекта на картинке просто приведет к смещению активации на карте признаков, модель все равно распознает паттерн. Это делает CNN устойчивой к небольшим смещениям объектов.

**Результат:** CNN требует на несколько порядков меньше параметров при работе с изображениями и демонстрирует гораздо лучшую обобщающую способность (generalization).

# Блок: Сверточные нейронные сети (CNN)

---

## Часть 1: Суть CNN

### Вопрос 1: Какие основные операции составляют ядро CNN?

**Ответ:** CNN состоит из четырех основных операций, которые последовательно применяются к входному изображению:

#### 1. Операция свертки (Convolution)

Свертка — это применение фильтра (ядра) размером $k \times k$ к входному изображению. На каждой позиции вычисляется скалярное произведение фильтра и локального участка изображения.

**Формула свертки:**

Для входного слоя $I$ и фильтра $K$ размером $k \times k$, выход на позиции $(i, j)$:

$$O_{i,j} = \sum_{u=0}^{k-1} \sum_{v=0}^{k-1} I_{i+u, j+v} \cdot K_{u,v} + b$$

где $b$ — смещение (bias).

**Размер выходной карты признаков:**

Если входное изображение размером $H \times W$, используется фильтр размером $k \times k$ с padding $p$ и stride $s$, то размер выхода:

$$H_{out} = \frac{H + 2p - k}{s} + 1, \quad W_{out} = \frac{W + 2p - k}{s} + 1$$

#### 2. Нелинейность (Activation Function)

После свертки применяется функция активации (обычно ReLU) для введения нелинейности:

$$f(x) = \max(0, x)$$

#### 3. Объединение (Pooling)

Pooling уменьшает пространственные размеры карты признаков, сохраняя наиболее важную информацию. Два основных типа:

**Max Pooling:** На каждом окне берется максимальное значение:

$$O_{i,j} = \max_{u,v} I_{i \cdot s + u, j \cdot s + v}$$

где $s$ — stride, а размер окна обычно $2 \times 2$.

**Average Pooling:** На каждом окне берется среднее значение:

$$O_{i,j} = \frac{1}{k^2} \sum_{u=0}^{k-1} \sum_{v=0}^{k-1} I_{i \cdot s + u, j \cdot s + v}$$

#### 4. Полносвязные слои (Fully Connected Layers)

После нескольких слоев свертки и pooling карта признаков преобразуется в одномерный вектор (flattening), который подается на полносвязные слои для выполнения задачи классификации.

---

### Вопрос 2: Что такое рецептивное поле (Receptive Field) и почему оно важно?

**Ответ:** Рецептивное поле нейрона в CNN — это размер области входного изображения, которая влияет на этот нейрон.

**Формула рецептивного поля:**

Для слоя $l$ рецептивное поле вычисляется рекурсивно:

$$RF_l = RF_{l-1} + (k_l - 1) \cdot \prod_{i=1}^{l-1} s_i$$

где $k_l$ — размер фильтра на слое $l$, $s_i$ — stride на слое $i$.

**Почему это важно:**

- Нейроны в первых слоях имеют маленькое рецептивное поле и усваивают детали (edges, текстуры)
- Нейроны в глубоких слоях имеют огромное рецептивное поле и видят глобальные паттерны (объекты, лица)
- Увеличение рецептивного поля достигается через увеличение stride или глубины сети

---

## Часть 2: Архитектуры CNN

### Вопрос 3: Опишите основные архитектуры CNN и их ключевые отличия.

**Ответ:**

#### LeNet-5 (1998, Yann LeCun)

**Назначение:** Распознавание рукописных цифр (MNIST).

**Архитектура:**
```
Input (32×32) → Conv(6, 5×5) → Pool(2×2) → Conv(16, 5×5) → Pool(2×2) → FC(120) → FC(84) → FC(10)
```

**Характеристики:**
- Всего ~60,000 параметров
- Работает на CPU за миллисекунды
- Очень простая по современным меркам

---

#### AlexNet (2012, Krizhevsky, Sutskever, Hinton)

**Назначение:** Классификация на ImageNet (1000 классов).

**Архитектура:**
```
Input (227×227×3) → Conv(96, 11×11, stride=4) → Pool(3×3, stride=2) → 
Conv(256, 5×5) → Pool(3×3) → Conv(384, 3×3) → Conv(384, 3×3) → Conv(256, 3×3) → 
Pool(3×3) → FC(4096) → Dropout(0.5) → FC(4096) → Dropout(0.5) → FC(1000)
```

**Ключевые инновации:**
- Использование ReLU вместо Sigmoid (обучается в 6 раз быстрее)
- Dropout для борьбы с переобучением
- GPU-параллелизм (обучалась на 2 GPU)
- Data augmentation
- ~60,000,000 параметров

---

#### VGG-16 (2014, Simonyan & Zisserman)

**Назначение:** Классификация на ImageNet.

**Архитектура:**
- Использует только фильтры размером $3 \times 3$
- Несколько фильтров $3 \times 3$ подряд дают больше нелинейности и больше рецептивное поле, чем один большой фильтр
- Глубокая и простая архитектура

```
Input → [Conv(64, 3×3) × 2] → Pool → [Conv(128, 3×3) × 2] → Pool → 
[Conv(256, 3×3) × 3] → Pool → [Conv(512, 3×3) × 3] → Pool → 
[Conv(512, 3×3) × 3] → Pool → FC(4096) → FC(4096) → FC(1000)
```

**Характеристики:**
- ~138,000,000 параметров (очень большая!)
- Очень хорошо интерпретируема
- Требует много памяти при обучении

---

#### ResNet (2015, He et al.)

**Назначение:** Классификация на ImageNet, победитель ILSVRC 2015.

**Архитектура:** Использует остаточные связи (residual connections / skip connections).

**Остаточный блок:**

```
x → Conv(64, 3×3) → BN → ReLU → Conv(64, 3×3) → BN → (+) ← x
```

**Формула:**

$$y = F(x) + x$$

где $F(x)$ — функция, которую нейросеть учится (обычно 2-3 свертки).

**Ключевые инновации:**
- Skip connections позволяют обучать намного более глубокие сети (до 152 слоев)
- Решает проблему затухающего градиента
- Batch Normalization после каждой свертки

---

#### MobileNet (2017, Howard et al.)

**Назначение:** Классификация на мобильных и встроенных устройствах.

**Ключевая техника: Depthwise Separable Convolution**

Вместо обычной свертки, которая применяет $K$ фильтров размером $k \times k \times C$ (где $C$ — количество каналов входа), используется:

1. **Depthwise Convolution:** $C$ фильтров размером $k \times k \times 1$ (один на каждый канал входа)
2. **Pointwise Convolution:** $K$ фильтров размером $1 \times 1 \times C$ (для комбинации каналов)

**Сложность:**

Обычная свертка: $H \cdot W \cdot K \cdot C \cdot k^2$ операций

Depthwise Separable: $H \cdot W \cdot C \cdot k^2 + H \cdot W \cdot K \cdot C$ операций

При $k = 3$ это снижает вычисления в ~8-9 раз!

---

## Часть 3: Параметры CNN

### Вопрос 4: Объясните ключевые гиперпараметры CNN и как они влияют на обучение.

**Ответ:**

#### 1. Размер фильтра (Filter Size / Kernel Size)

**Определение:** Размер окна свертки (обычно $3 \times 3$, $5 \times 5$, $7 \times 7$).

**Влияние:**

- **Маленькие фильтры ($3 \times 3$):** Захватывают локальные детали, работают лучше для края и текстуры
- **Большие фильтры ($7 \times 7$):** Захватывают более глобальный контекст, требуют больше параметров

**Правило:** Несколько маленьких фильтров ($k=3$) подряд лучше, чем один большой ($k=5$), так как:
- Меньше параметров
- Больше нелинейности
- Тот же рецептивное поле

---

#### 2. Количество фильтров (Number of Filters / Channels)

**Определение:** Сколько фильтров применяется на каждом слое.

**Влияние:**

- Первые слои: 32-64 фильтра (нужно мало для простых паттернов: edges)
- Средние слои: 128-256 фильтров (улавливают сложные паттерны)
- Глубокие слои: 512 фильтров (абстрактные признаки)

**Правило:** Обычно количество фильтров увеличивается с глубиной, так как карта признаков становится меньше пространственно.

---

#### 3. Padding

**Определение:** Добавление нулей или других значений по краям входного изображения.

**Две стратегии:**

- **'valid' или padding=0:** Нет padding, размер выхода уменьшается
  $$H_{out} = \frac{H - k}{s} + 1$$

- **'same' или padding='same':** Padding такой, чтобы при stride=1 размер остался прежним
  $$p = \frac{k - 1}{2}$$
  $$H_{out} = H$$

**Влияние:**

- Без padding: информация на краях изображения используется реже (градиент там слабее)
- С padding: краны обрабатываются одинаково с центром

**Правило:** Обычно используется **'same' padding** для сохранения глубины пространственных измерений.

---

#### 4. Stride (Шаг свертки)

**Определение:** На сколько пикселей сдвигается фильтр за раз.

**Частые значения:**
- stride=1: Фильтр двигается на 1 пиксель (стандартно)
- stride=2: Фильтр двигается на 2 пикселя (часто первый слой в ResNet, MobileNet)

**Формула:** При stride=s размер выхода:

$$H_{out} = \frac{H + 2p - k}{s} + 1$$

**Влияние:**

- stride=1: Выхлопывает детали, но требует больше вычислений
- stride=2: Быстро уменьшает размер, может потерять информацию

**Правило:** Обычно используется stride=1 в сверточных слоях, stride=2 в pooling или первом слое для быстрого снижения размерности.

---

#### 5. Pooling Size и Stride

**Max Pooling vs Average Pooling:**

**Max Pooling (стандартно):**
- Берет максимальное значение из окна
- Лучше для классификации
- Более "острая" информация

**Average Pooling:**
- Берет среднее значение из окна
- Гладче, менее чувствителен к шуму
- Реже используется в классификации

**Размер pooling окна:**

Обычно $2 \times 2$ с stride=2:

```
Input: 28×28 → MaxPool(2×2, stride=2) → Output: 14×14
```

**Правило:** Pooling обычно следует за 2-3 свертками, чтобы постепенно уменьшать пространственный размер.

---

#### 6. Dropout

**Определение:** Во время обучения каждый нейрон "отключается" с вероятностью $p$ (обычно 0.5).

**Формула:** Во время обучения активации масштабируются:

$$\tilde{a} = \begin{cases} 
0 & \text{с вероятностью } p \\
\frac{a}{1-p} & \text{с вероятностью } 1-p
\end{cases}$$

Во время инференса dropout не применяется.

**Влияние:**

- Действует как ансамбль (как будто обучаем множество подсетей одновременно)
- Борется с переобучением
- Работает лучше, чем L1/L2 регуляризация для глубоких сетей

**Правило:** Обычно 0.5 для полносвязных слоев, 0.1-0.3 для сверточных (меньше переобучение в сверточных).

---

### Вопрос 5: Как структурировать CNN практически? (Best Practices)

**Ответ:**

#### Общая схема проектирования:

```
Input (3×224×224)
  ↓
[Stem Block] - быстрое снижение размерности
  - Conv(64, 7×7, stride=2) или Conv(32, 3×3, stride=2) + Conv(64, 3×3, stride=2)
  - BatchNorm → ReLU
  - MaxPool(3×3, stride=2)
  - Output: (64×56×56)
  ↓
[Main Blocks] - 3-4 блока с расширением каналов и снижением размера
  - Блок 1: 64 каналов, 4-6 остаточных связей, output (64×56×56)
  - Блок 2: 128 каналов, 4-6 остаточных связей, output (128×28×28)
  - Блок 3: 256 каналов, 4-6 остаточных связей, output (256×14×14)
  - Блок 4: 512 каналов, 4-6 остаточных связей, output (512×7×7)
  ↓
[Global Average Pooling]
  - Усредняет пространственные измерения
  - Output: (512,)
  ↓
[Classification Head]
  - FC(num_classes)
  - Output: (num_classes,)
```

#### Правила выбора параметров:

| Слой | Фильтр | Stride | Padding | Примечание |
|------|--------|--------|---------|-----------|
| Первый сверточный | 7×7 | 2 | 3 | Быстрое снижение размера на 4x |
| Основные свертки | 3×3 | 1 | 1 | Локальные патерны, часто сложены по 2-3 |
| После свертки | - | 2 | 1 | MaxPool(2×2, stride=2) для дальнейшего снижения |
| Depthwise Conv | 3×3 | 1-2 | 1 | Для мобильных архитектур |

#### Количество параметров по слоям:

Для Conv слоя: $K \cdot C_{in} \cdot k^2 + K$ (включая bias)

где $K$ — количество output фильтров, $C_{in}$ — количество input каналов, $k$ — размер фильтра.

Пример для Conv(128, 64, 3×3):
$$128 \cdot 64 \cdot 3^2 + 128 = 73,856 \text{ параметров}$$

# Блок: Компьютерное зрение (Computer Vision)

---

## Часть 1: Типы задач в Computer Vision

### Вопрос 1: Какие основные типы задач решает Computer Vision?

**Ответ:** Computer Vision включает множество различных задач, которые можно категоризировать по сложности и типу выхода:

#### 1. Классификация изображений (Image Classification)

**Задача:** Определить, какой класс/категория изображена на всей картинке.

**Вход:** Изображение $I$ размером $H \times W \times 3$

**Выход:** Вероятность каждого класса: $P(c_k | I)$ для каждого класса $k$

**Примеры:**
- Распознавание животных
- Медицинское диагностирование (рак/норма)
- Классификация автомобилей по марке

**Метрики:** Accuracy, Top-5 Accuracy, ROC-AUC

---

#### 2. Локализация объектов (Object Localization)

**Задача:** Найти класс объекта и его координаты на изображении (обычно один главный объект).

**Выход:** Класс + bounding box $(x_{min}, y_{min}, x_{max}, y_{max})$

**Примеры:**
- Поиск человека на фото
- Поиск машины на дорожной камере

**Отличие от классификации:** Добавляется координатная информация.

---

#### 3. Детекция объектов (Object Detection)

**Задача:** Найти все объекты на изображении, определить их классы и координаты.

**Выход:** Множество bounding boxes с классами:
$$\{(c_i, x_i, y_i, w_i, h_i)\} \text{ для } i = 1..N$$

где $c_i$ — класс, $(x_i, y_i)$ — центр, $w_i, h_i$ — ширина и высота.

**Примеры:**
- Детекция лиц на фото
- Обнаружение пешеходов на дороге
- Поиск объектов в видео в реальном времени

**Метрики:** mAP (Mean Average Precision), IoU (Intersection over Union)

---

#### 4. Семантическая сегментация (Semantic Segmentation)

**Задача:** Для каждого пикселя определить, к какому классу он принадлежит.

**Выход:** Маска размером $H \times W$, где каждый пиксель имеет метку класса.

**Примеры:**
- Выделение дороги, деревьев, людей на снимке города
- Медицинская сегментация органов на МРТ
- Сегментация типов клеток на микроскопии

**Характеристика:** Не различает разные объекты одного класса.

---

#### 5. Экземплярная сегментация (Instance Segmentation)

**Задача:** Для каждого пикселя определить класс и конкретный объект (экземпляр).

**Выход:** Маска размером $H \times W \times ID$, где каждый объект имеет уникальный ID.

**Примеры:**
- Выделение каждого человека отдельно
- Выделение каждой машины отдельно
- Подсчет популяции животных

**Отличие от семантической:** Разделяет объекты одного класса на отдельные экземпляры.

---

#### 6. Ключевые точки и поза (Pose Estimation & Keypoint Detection)

**Задача:** Найти ключевые точки на объекте (суставы на теле, черты лица и т.д.).

**Выход:** Координаты $N$ точек: $\{(x_i, y_i, confidence_i)\}$

**Примеры:**
- Определение позы человека
- Распознавание лицевых ориентиров
- Отслеживание движений животных

---

#### 7. Отслеживание объектов (Object Tracking)

**Задача:** Отслеживать один или несколько объектов через кадры видео.

**Выход:** Для каждого кадра $t$ — координаты объекта $(x_t, y_t, w_t, h_t)$

**Примеры:**
- Отслеживание мяча в спорте
- Отслеживание пешехода на улице
- Отслеживание автомобиля через несколько кадров

**Метрики:** MOT (Multiple Object Tracking) accuracy

---

| Задача | Выход | Сложность | Основной алгоритм |
|---|---|---|---|
| Классификация | Класс | ⭐ | CNN, VGG, ResNet |
| Локализация | Класс + BBox | ⭐⭐ | R-CNN variants |
| Детекция | Множество BBox | ⭐⭐⭐ | YOLO, Faster R-CNN |
| Сегментация семантическая | Маска пикселей | ⭐⭐⭐ | U-Net, FCN |
| Сегментация экземплярная | Маска + ID | ⭐⭐⭐⭐ | Mask R-CNN |
| Поза | Ключевые точки | ⭐⭐⭐ | OpenPose, COCO |
| Отслеживание | BBox per frame | ⭐⭐⭐⭐ | Kalman Filter + CNN |

## Часть 3: YOLO (You Only Look Once)

### Вопрос 3: Краткая история YOLO и почему это была революция?

**Ответ:** До YOLO рынок детекции был в основном за двухстадийными моделями (например, Faster R-CNN):

1. Сначала ищутся region proposals
2. Потом каждый регион классифицируется

Это давало высокую точность, но большую задержку. YOLO предложила одношаговую идею: сеть смотрит на изображение один раз и сразу выдает координаты боксов и классы. Детекция стала задачей регрессии end-to-end.

### Эпоха 1: Отцы-основатели (Joseph Redmon)

#### YOLOv1 (2015)
- Рождение одношаговой детекции
- Разбиение изображения на сетку (например, $7 \times 7$)
- Каждая ячейка предсказывает bounding boxes и классы
- Прорыв по скорости, но слабее на мелких объектах и плотных сценах

#### YOLOv2 (2016)
- Batch Normalization
- Anchor Boxes
- Multi-scale training
- Более стабильное обучение и рост точности

#### YOLOv3 (2018)
- FPN-подход: предсказания на нескольких масштабах
- Backbone Darknet-53
- Значительно лучшее качество на мелких объектах
- После этого релиза Redmon ушел из CV-исследований

### Эпоха 2: Расцвет сообщества и форков

#### YOLOv4 (2020)
- Работа Alexey Bochkovskiy
- Bag of Freebies (например, Mosaic) и Bag of Specials
- CSPDarknet как backbone
- Большой прирост practical accuracy без потери скорости

#### YOLOv5 (2020, Ultralytics)
- Инженерный прорыв: удобный PyTorch-стек
- Улучшенный train/deploy pipeline
- Семейство размеров моделей для разных устройств
- Очень широкое adoption в production

#### YOLOv6 (2022) и YOLOv7 (2022)
- Фокус на hardware-friendly оптимизации
- Улучшение прохождения градиентов и эффективности инференса

### Эпоха 3: Anchor-free и унификация задач

#### YOLOv8 (2023, Ultralytics)
- Переход к anchor-free логике
- Единый фреймворк для detection, segmentation, pose, classification, tracking
- Более простой и стабильный pipeline обучения и экспорта

#### YOLOv9 (2024)
- Programmable Gradient Information (PGI)
- Улучшение передачи полезной информации в глубокой сети

#### YOLOv10 (2024)
- NMS-free направление
- Снижение latency за счет end-to-end детекции без тяжелого постпроцессинга

#### YOLO11 (2024)
- Дальнейшая оптимизация блоков
- Улучшения для прикладных сценариев, включая oriented bounding boxes

### Эпоха 4: Edge-революция (2025-2026)

#### YOLOv12 (2025)
- Больше внимания к контексту сцены (area attention идеи)
- Усиление качества на сложных сценах и во временной динамике

#### YOLO26 (конец 2025 / 2026)
- Крупный редизайн под edge-сценарии: мобильные платформы, дроны, робототехника, IoT
- Фокус на минимальную задержку и стабильный инференс на слабых чипах

**Ключевые заявленные акценты YOLO26:**
- End-to-end без NMS как целевая архитектурная идея
- Упрощение экспортного пути (ONNX/TensorRT)
- Более стабильная оптимизация лосса для сложных датасетов и мелких объектов

### Почему YOLO остается ключевой в 2026

- Лучший практический баланс speed/accuracy для real-time задач
- Широкий ecosystem и готовые production-инструменты
- Быстрый цикл обновлений и хорошая поддержка edge-девайсов

---

### Вопрос 4: Как работает YOLO на практическом уровне?

**Ответ:**

1. Изображение подается в backbone и преобразуется в multi-scale признаки
2. Head предсказывает bounding boxes, confidence и классы
3. Для финального ответа применяется постобработка (или end-to-end логика в новых версиях)

**Типичная формула confidence (классическая постановка):**
$$Confidence = P(Object) \cdot IoU_{pred}^{truth}$$

### NMS (классический постпроцессинг)

1. Отсортировать боксы по confidence
2. Взять бокс с максимальным confidence
3. Удалить боксы с высоким IoU к выбранному
4. Повторить

**Результат:** Удаляются дубликаты и остается финальный набор детекций.

---

### Пример кода YOLO (Ultralytics)

```python
from ultralytics import YOLO

model = YOLO('yolov8m.pt')
results = model.predict(source='image.jpg', conf=0.25)

for det in results[0].boxes:
    x1, y1, x2, y2 = det.xyxy[0]
    conf = det.conf[0]
    cls = int(det.cls[0])
    print(model.names[cls], float(conf), (float(x1), float(y1), float(x2), float(y2)))
```

# Блок: NLP (Natural Language Processing) — базовые знания

---

## Вопрос 1: Что такое NLP и какие задачи он решает?

**Ответ:** NLP (обработка естественного языка) — это область машинного обучения, где модели работают с человеческим языком: текстом и речью.

**Основные задачи NLP:**
- Классификация текста (спам/не спам, тональность)
- Последовательная разметка (NER: имена, компании, даты)
- Машинный перевод
- Суммаризация
- Вопрос-ответ
- Поиск и ранжирование документов
- Генерация текста (чат-боты, ассистенты)

---

## Вопрос 2: Из каких этапов состоит базовый NLP-пайплайн?

**Ответ:**

1. **Сбор и очистка данных**
Удаление HTML, лишних символов, дубликатов, приведение кодировок.

2. **Нормализация текста**
Lowercase (если нужно), токенизация, лемматизация/стемминг, работа со стоп-словами.

3. **Преобразование текста в числа**
Векторайзеры: BoW, TF-IDF, Hashing, Word2Vec/FastText, BERT-эмбеддинги.

4. **Обучение модели**
Выбор алгоритма: логистическая регрессия, SVM, LSTM, Transformer и т.д.

5. **Оценка качества**
Метрики: Accuracy, Precision, Recall, F1, ROC-AUC, BLEU/ROUGE (для генерации/перевода).

6. **Деплой и мониторинг**
Слежение за деградацией качества (data drift, concept drift).

---

## Вопрос 3: Что такое токенизация, стемминг и лемматизация?

**Ответ:**
- **Токенизация** — разбиение текста на части (слова, подслова, символы).
- **Стемминг** — грубое отсечение окончаний (быстро, но иногда неточно).
- **Лемматизация** — приведение слова к словарной форме (точнее, но тяжелее).

**Пример:**
- "бежали", "бегу", "бегает"
- Стемминг: "бег"
- Лемматизация: "бежать"

---

## Вопрос 4: Какие бывают векторайзеры текста и как они работают?

**Ответ:** Векторайзер превращает текст в числовой вектор, с которым уже может работать модель. Основные семейства:

### 1. CountVectorizer (Bag of Words)

**Идея:** считаем, сколько раз каждое слово встретилось в документе.

Если словарь размера $V$, документ — это вектор $x \in \mathbb{R}^{V}$, где:

$$x_j = \text{count}(t_j, d)$$

**Плюсы:**
- Очень быстро
- Просто интерпретировать
- Хороший baseline

**Минусы:**
- Не учитывает смысл слова
- Очень разреженные векторы
- Неустойчив к синонимам

---

### 2. TF-IDF Vectorizer

**Идея:** важность слова = частота в документе * редкость в корпусе.

$$TF(t, d) = \frac{f_{t,d}}{\sum_{t'} f_{t',d}}$$
$$IDF(t) = \log\frac{N}{df(t)}$$
$$TF\text{-}IDF(t, d) = TF(t,d) \cdot IDF(t)$$

где:
- $f_{t,d}$ — число вхождений термина $t$ в документ $d$
- $N$ — число документов
- $df(t)$ — в скольких документах встречается термин $t$

**Плюсы:**
- Лучше BoW для классификации и поиска
- Подавляет служебные часто встречающиеся слова
- Отлично работает в малых/средних задачах

**Минусы:**
- Все еще без контекста
- Разреженные признаки

---

### 3. HashingVectorizer

**Идея:** вместо явного словаря токен хэшируется в один из $m$ индексов.

$$index = h(token) \bmod m$$

**Плюсы:**
- Не хранит словарь (экономит память)
- Подходит для stream/online сценариев
- Быстрый и масштабируемый

**Минусы:**
- Коллизии (разные слова могут попадать в один индекс)
- Хуже интерпретируемость

---

### 4. Word2Vec

**Идея:** обучить плотные векторы слов так, чтобы слова в похожем контексте имели близкие векторы.

Два режима:
- **CBOW:** предсказывает текущее слово по соседям
- **Skip-gram:** предсказывает соседей по текущему слову

**Плюсы:**
- Плотные компактные векторы (например, 100-300)
- Отражают семантику

**Минусы:**
- Один вектор на слово, без учета контекста
- Плохо для омонимов

---

### 5. FastText

**Идея:** как Word2Vec, но с учетом символьных n-грамм (подслов).

Слово представляется как сумма векторов n-грамм:

$$v(word) = \sum_{g \in G(word)} v_g$$

**Плюсы:**
- Лучше работает с редкими словами и опечатками
- Хорош для морфологически богатых языков (включая русский)

**Минусы:**
- Все еще статические эмбеддинги

---

### 6. Doc2Vec / усреднение эмбеддингов

**Идея:** получить вектор всего документа:
- либо отдельной моделью (Doc2Vec)
- либо усреднением векторов слов

$$v(doc) = \frac{1}{n} \sum_{i=1}^{n} v(w_i)$$

**Плюсы:**
- Простой переход к задачам классификации документа

**Минусы:**
- Усреднение теряет порядок слов и синтаксис

---

### 7. Контекстные эмбеддинги (BERT, RoBERTa, sentence-transformers)

**Идея:** вектор слова/предложения зависит от контекста.

Слово "лук" в разных предложениях получает разные векторы.

**Плюсы:**
- Лучшее качество на большинстве NLP-задач
- Учитывается контекст, порядок и взаимодействия слов

**Минусы:**
- Более тяжелые вычисления
- Требуют GPU для массового обучения/инференса

---

## Вопрос 5: Как выбрать векторайзер на практике? (Что когда лучше)

**Ответ:**

### Быстрый выбор

- **Мало данных + нужна интерпретируемость:** TF-IDF + Logistic Regression/SVM
- **Большие потоковые данные и ограниченная память:** HashingVectorizer
- **Классическая семантика без тяжелых моделей:** FastText
- **Максимальное качество на современных задачах:** BERT/sentence-transformers
- **Семантический поиск и RAG:** sentence-transformers (эмбеддинги предложений) + векторная БД

### Под задачу

- **Классификация отзывов/тикетов:**
1. baseline: TF-IDF
2. если качество не хватает: BERT fine-tuning

- **Поиск похожих документов:**
1. TF-IDF/BM25 как базовый ранжировщик
2. dense retrieval на sentence embeddings как улучшение

- **Работа с шумным текстом, опечатками, морфологией:** FastText

- **Жесткие ограничения по latency на CPU:** TF-IDF/Hashing

### Под размер данных

- **До ~50k документов:** TF-IDF почти всегда отличный старт
- **50k - 1M:** TF-IDF + линейные модели или FastText
- **1M+ и высокие требования к качеству:** гибрид sparse+dense, трансформеры

## Вопрос 6: Чем sparse и dense представления отличаются математически?

**Ответ:**

- **Sparse (BoW, TF-IDF):**
  Вектор большой размерности $V$, но почти все координаты равны нулю.
  Похожесть часто считается косинусом:
  $$\cos(\theta) = \frac{x \cdot y}{\|x\|\|y\|}$$

- **Dense (эмбеддинги):**
  Вектор малой размерности $d$ (например, 384/768), почти все координаты ненулевые.
  Похожесть: cosine similarity или dot product.

**Практический смысл:**
- Sparse лучше ловит точные совпадения слов
- Dense лучше ловит смысловую близость
- В production часто используют гибрид: sparse + dense

- ограничены вычислительные ресурсы

# Блок: RNN, LSTM, GRU, BiLSTM — подробно

---

## Вопрос 1: Что такое RNN и в чем идея архитектуры?

**Ответ:** RNN (Recurrent Neural Network) — это класс нейросетей для последовательностей, где состояние на шаге $t$ зависит от текущего входа и прошлого состояния.

Базовые уравнения:

$$h_t = \phi(W_{xh}x_t + W_{hh}h_{t-1} + b_h)$$
$$y_t = W_{hy}h_t + b_y$$

где:
- $x_t$ — входной токен на шаге $t$
- $h_t$ — скрытое состояние (память)
- $\phi$ — нелинейность (обычно tanh)

**Интуиция:** модель читает последовательность слева направо и переносит «контекст» через $h_t$.

**Где применяли:** классификация текста, NER, машинный перевод, прогноз временных рядов, распознавание речи.

---

## Вопрос 2: В чем ключевая проблема обычной RNN?

**Ответ:** затухающие и взрывающиеся градиенты при Backpropagation Through Time (BPTT).

- При длинных последовательностях влияние ранних токенов теряется
- Трудно выучить долгосрочные зависимости
- Обучение становится нестабильным

Это главная причина, почему появились LSTM и GRU.

---

## Вопрос 3: Как устроена LSTM?

**Ответ:** LSTM добавляет отдельную ячейку памяти $c_t$ и систему ворот, которые управляют тем, что забыть, что записать и что отдать наружу.

Уравнения LSTM:

$$f_t = \sigma(W_f[x_t, h_{t-1}] + b_f) \quad \text{(forget gate)}$$
$$i_t = \sigma(W_i[x_t, h_{t-1}] + b_i) \quad \text{(input gate)}$$
$$\tilde{c}_t = \tanh(W_c[x_t, h_{t-1}] + b_c)$$
$$c_t = f_t \odot c_{t-1} + i_t \odot \tilde{c}_t$$
$$o_t = \sigma(W_o[x_t, h_{t-1}] + b_o) \quad \text{(output gate)}$$
$$h_t = o_t \odot \tanh(c_t)$$

**Почему лучше обычной RNN:**
- Лучше держит долгую память
- Стабильнее в обучении
- Меньше страдает от затухания градиента

**Минусы:**
- Много параметров
- Более медленный шаг по сравнению с GRU

---

## Вопрос 4: Как устроена GRU?

**Ответ:** GRU (Gated Recurrent Unit) — упрощенная версия LSTM: без отдельного состояния ячейки $c_t$, только скрытое состояние $h_t$ и два типа ворот.

Уравнения GRU:

$$z_t = \sigma(W_z[x_t, h_{t-1}] + b_z) \quad \text{(update gate)}$$
$$r_t = \sigma(W_r[x_t, h_{t-1}] + b_r) \quad \text{(reset gate)}$$
$$\tilde{h}_t = \tanh(W_h[x_t, r_t \odot h_{t-1}] + b_h)$$
$$h_t = (1-z_t) \odot h_{t-1} + z_t \odot \tilde{h}_t$$

**Плюсы GRU:**
- Меньше параметров, чем у LSTM
- Быстрее обучается
- Часто дает сопоставимое качество

**Минусы GRU:**
- Иногда уступает LSTM на задачах с очень длинной зависимостью
- Чуть меньше контроля над «памятью», чем у LSTM

---

## Вопрос 5: Что такое BiLSTM и зачем он нужен?

**Ответ:** BiLSTM (Bidirectional LSTM) — это две LSTM:
- одна читает последовательность слева направо
- вторая справа налево

Далее состояния объединяются (concat/sum).

$$h_t^{bi} = [\overrightarrow{h_t}; \overleftarrow{h_t}]$$

**Что дает:**
- Учитывается левый и правый контекст одновременно
- Особенно полезно в sequence labeling (NER, POS-tagging)

**Ограничение:**
- Нужна вся последовательность целиком, поэтому сложнее для real-time стриминга
- Дороже по вычислениям (фактически два прохода)

---

## Вопрос 6: LSTM vs GRU vs BiLSTM — что выбрать?

**Короткий практический ориентир:**

- **GRU**: когда нужен хороший компромисс скорость/качество
- **LSTM**: когда важна более мощная память на длинных последовательностях
- **BiLSTM**: когда важен контекст с обеих сторон и нет жесткого online-ограничения

**Частый baseline для классических задач NLP:** Embedding -> BiLSTM -> Linear/CRF.

---

## Вопрос 7: Почему RNN/LSTM/GRU/BiLSTM считаются дорогими?

**Ответ:** дорого и по времени, и по памяти.

### 1. Последовательная природа вычислений

Для шага $t$ нужен результат шага $t-1$. Это значит, что по времени токены почти нельзя полноценно параллелить.

- У Transformer большая часть вычислений параллелится по токенам
- У RNN семейства последовательность является «бутылочным горлышком»

### 2. BPTT хранит много промежуточных состояний

При обучении нужно сохранить активации для всех шагов, чтобы потом посчитать градиенты назад по времени.

Память примерно растет как:

$$O(B \cdot T \cdot H)$$

где:
- $B$ — batch size
- $T$ — длина последовательности
- $H$ — размер скрытого состояния

### 3. Воротные механизмы увеличивают стоимость шага

LSTM и GRU содержат несколько матричных операций на каждом шаге.
- LSTM обычно тяжелее GRU
- BiLSTM еще дороже, потому что два направления

### 4. Длинные последовательности особенно затратны

Чем длиннее $T$, тем:
- больше шагов в forward
- длиннее backward
- выше риск нестабильности и дольше обучение

---

## Вопрос 8: Как удешевить обучение RNN на практике?

- Использовать **GRU** вместо LSTM, если качество не падает
- Ограничивать длину последовательности (truncation, bucketing, padding до разумного max_len)
- Применять **gradient clipping** для стабильности
- Использовать mixed precision
- Уменьшать hidden size и число слоев
- Для очень длинного контекста рассмотреть переход на Transformer-архитектуры

---

## Мини-резюме

1. Обычная RNN проста, но плохо держит длинную память
2. LSTM добавляет память и ворота, обычно надежнее на длинных зависимостях
3. GRU проще и быстрее, часто почти не проигрывает
4. BiLSTM дает контекст с двух сторон, но стоит дороже
5. Дорого из-за последовательности вычислений, BPTT и большого числа операций на шаг

# Блок: Transformers и Attention — ключевые идеи

---

## Вопрос 1: Почему вообще понадобились Transformers?

**Ответ:** до трансформеров в NLP доминировали RNN/LSTM/GRU, но у них есть системные ограничения:

- вычисления идут последовательно по времени ($t-1 \to t$), плохо параллелятся
- на длинных последовательностях сложнее держать зависимости
- обучение медленнее, особенно при больших корпусах

**Transformer** решает это через механизм **self-attention**:
- каждый токен может сразу "посмотреть" на все остальные токены
- вычисления по токенам хорошо параллелятся на GPU
- лучше захватываются дальние зависимости

---

## Вопрос 2: Что такое Attention на интуитивном уровне?

**Ответ:** Attention отвечает на вопрос: "какие слова в последовательности наиболее важны для текущего слова/позиции?"

Пример: в фразе "Кошка сидела на ковре, потому что она устала" слово "она" должно сильнее смотреть на "кошка", чем на "ковре".

То есть модель не одинаково учитывает весь контекст, а назначает **веса важности**.

---

## Вопрос 3: Как работает Scaled Dot-Product Attention (математика)?

У каждого токена строятся три вектора:
- **Q (Query)**: что я ищу
- **K (Key)**: что я могу предложить
- **V (Value)**: какую информацию я передаю

Для матриц $Q, K, V$:

$$\mathrm{Attention}(Q, K, V) = \mathrm{softmax}\left(\frac{QK^T}{\sqrt{d_k}}\right)V$$

Разбор формулы:
- $QK^T$ дает "оценки совместимости" между токенами
- деление на $\sqrt{d_k}$ стабилизирует масштаб логитов
- softmax превращает оценки в вероятностные веса
- взвешенная сумма $V$ дает новый контекстный вектор

**Почему это сильно:** каждый токен получает представление, зависящее от всех релевантных токенов сразу.

---

## Вопрос 4: Что такое Self-Attention?

**Ответ:** это случай attention, когда $Q, K, V$ берутся из одной и той же последовательности.

- Вход: эмбеддинги токенов $X \in \mathbb{R}^{n \times d}$
- Линейные проекции:

$$Q = XW_Q, \quad K = XW_K, \quad V = XW_V$$

- Затем применяется attention-формула.

**Результат:** новый вектор токена учитывает контекст всей последовательности.

---

## Вопрос 5: Зачем нужен Multi-Head Attention?

**Ответ:** одна "голова" attention может фокусироваться на одном типе связей, но язык многослоен. Поэтому делают несколько голов параллельно.

Для каждой головы $i$:

$$\text{head}_i = \mathrm{Attention}(Q_i, K_i, V_i)$$

После этого:

$$\mathrm{MultiHead}(Q,K,V) = \mathrm{Concat}(\text{head}_1, ..., \text{head}_h)W_O$$

**Интуиция:**
- одна голова ловит синтаксис
- другая coreference (ссылки местоимений)
- третья дальние тематические связи

Это повышает выразительность без радикального усложнения архитектуры.

---

## Вопрос 6: Из чего состоит Transformer-блок?

Классический блок (Vaswani et al.) содержит:

1. **Multi-Head Self-Attention**
2. **Add & Norm** (residual connection + LayerNorm)
3. **Position-wise Feed-Forward Network (FFN)**
4. **Add & Norm**

FFN обычно:

$$\mathrm{FFN}(x) = W_2\,\sigma(W_1x + b_1) + b_2$$

где $\sigma$ часто ReLU или GELU.

**Почему residual + norm важны:**
- стабилизируют обучение
- упрощают оптимизацию глубоких стеков
- улучшают прохождение градиента

---

## Вопрос 7: Если в Transformer нет рекуррентности, как он понимает порядок слов?

**Ответ:** через **позиционные кодировки** (Positional Encoding).

Базовый синусоидальный вариант:

$$PE_{(pos,2i)} = \sin\left(\frac{pos}{10000^{2i/d}}\right), \quad
PE_{(pos,2i+1)} = \cos\left(\frac{pos}{10000^{2i/d}}\right)$$

Эти векторы добавляют к token embeddings:

$$X_{input} = E_{token} + E_{position}$$

Смысл: модель получает информацию не только "что за токен", но и "где он стоит".

---

## Вопрос 8: Encoder и Decoder в Transformer — в чем разница?

### Encoder
- стек одинаковых encoder-блоков
- внутри self-attention без причинной маски
- каждый токен видит весь вход

### Decoder
- masked self-attention (нельзя смотреть в будущее)
- cross-attention к выходам encoder
- используется для генерации

Идея seq2seq:
- encoder кодирует исходный текст
- decoder по одному токену генерирует ответ

---

## Вопрос 9: Что такое mask в attention и зачем она нужна?

**Ответ:** mask запрещает внимание к определенным позициям.

Основные типы:

- **Padding mask**: скрывает паддинг-токены
- **Causal mask** (look-ahead): в декодере токен на позиции $t$ не видит позиции $>t$

Практически это делается добавлением больших отрицательных значений к запрещенным логитам перед softmax.

---

## Вопрос 10: Почему Transformers так эффективны на длинных зависимостях?

**Ответ:** путь взаимодействия между двумя токенами в self-attention почти "один шаг".

Для RNN зависимость между далекими позициями проходит через длинную цепочку состояний. В Transformer любой токен может напрямую взвесить любой другой токен в том же слое.

Это помогает лучше учить:
- длинные логические связи
- дальние референции
- сложный контекст документа

---

## Вопрос 11: Сколько это стоит вычислительно?

### Сильная сторона
- Отличная параллелизация по токенам

### Узкое место
- У классического self-attention сложность по длине последовательности примерно:

$$O(n^2 \cdot d)$$

и память для attention-матрицы тоже растет как $O(n^2)$.

Если $n$ большой (длинные документы), это становится дорого по VRAM и времени.

Поэтому появились efficient attention-подходы (Longformer, Performer, FlashAttention и др.).

---

## Вопрос 12: Как выглядит полный поток данных в Transformer (очень кратко)?

1. Текст -> токенизация
2. Токены -> embeddings + positional encodings
3. Проход через несколько Transformer-блоков
4. Head под задачу:
- классификация (linear + softmax)
- токен-классификация (NER)
- генерация следующего токена

---

## Вопрос 13: Практические плюсы и минусы Transformer

**Плюсы:**
- сильное качество на большинстве NLP-задач
- хорошо масштабируется с данными и параметрами
- легко дообучать (fine-tuning, instruction tuning)

**Минусы:**
- дорого по памяти и вычислениям
- чувствительно к длине контекста
- требует аккуратной инфраструктуры (GPU/TPU, оптимизация инференса)

---

## Вопрос 14: Где внимание особенно важно в реальных задачах?

- Машинный перевод
- Суммаризация
- QA и RAG
- Семантический поиск
- Извлечение информации из длинных документов
- Генерация текста и диалоговые системы

---

## Мини-резюме по блоку

1. Attention позволяет модели избирательно фокусироваться на релевантных токенах
2. Self-attention делает контекст глобальным внутри слоя
3. Multi-head attention учит разные типы связей параллельно
4. Transformer = attention + FFN + residual + normalization + positional encoding
5. Главный компромисс: высокое качество и параллелизм против $O(n^2)$ по длине контекста




# Блок: Автоэнкодеры (Autoencoders) — от базы до практики

---

## Вопрос 1: Что такое автоэнкодер?

**Ответ:** автоэнкодер — это нейросеть, которая учится восстанавливать входные данные из сжатого представления.

Базовая схема:

$$x \xrightarrow{Encoder} z \xrightarrow{Decoder} \hat{x}$$

где:
- $x$ — исходный объект (вектор, изображение, спектр, текстовое представление)
- $z$ — латентный вектор (bottleneck)
- $\hat{x}$ — реконструкция входа

Цель обучения:

$$\min \mathcal{L}(x, \hat{x})$$

То есть мы минимизируем ошибку реконструкции.

---

## Вопрос 2: Из каких частей состоит архитектура?

### 1. Encoder
Преобразует вход в компактный латентный код $z$.

### 2. Bottleneck (latent space)
Узкое место сети. Именно здесь модель вынуждена оставить только важную информацию.

### 3. Decoder
Пытается восстановить исходный объект из $z$.

Если bottleneck достаточно узкий, сеть учится полезным признакам, а не просто копированию.

---

## Вопрос 3: Какие бывают функции потерь у автоэнкодеров?

Зависит от типа данных:

- **MSE (L2):**
$$\mathcal{L}_{MSE} = \frac{1}{n}\sum_i (x_i-\hat{x}_i)^2$$

- **MAE (L1):**
$$\mathcal{L}_{MAE} = \frac{1}{n}\sum_i |x_i-\hat{x}_i|$$

- **Binary Cross-Entropy** (если пиксели/признаки нормированы в $[0,1]$)

Практически:
- для непрерывных признаков часто MSE
- для бинарных/вероятностных признаков BCE

---

## Вопрос 4: Зачем автоэнкодеры используют на практике?

- Снижение размерности (альтернатива PCA, но нелинейная)
- Шумоподавление (denoising)
- Поиск аномалий (аномалии реконструируются хуже)
- Извлечение признаков для downstream-моделей
- Сжатие представлений
- Предобучение (в некоторых пайплайнах)

---

## Вопрос 5: Основные типы автоэнкодеров

### 1. Vanilla Autoencoder
Обычный encoder-decoder, обучается на реконструкцию.

**Плюс:** простота.
**Минус:** может переобучаться и учить «почти identity», если bottleneck слишком широкий.

---

### 2. Sparse Autoencoder
Добавляют ограничение на активации (например, L1), чтобы большинство нейронов было "молчащими".

Идея лосса:

$$\mathcal{L} = \mathcal{L}_{recon} + \lambda \cdot \Omega(h)$$

где $\Omega(h)$ штрафует плотные активации.

**Плюс:** более информативные и разреженные признаки.

---

### 3. Denoising Autoencoder (DAE)
На вход дают зашумленный объект $\tilde{x}$, а восстанавливают чистый $x$.

$$\tilde{x} = x + \epsilon, \quad \min \mathcal{L}(x, Decoder(Encoder(\tilde{x})))$$

**Плюс:** устойчивость к шуму, лучшее обобщение.

---

### 4. Contractive Autoencoder
Штрафуют чувствительность латента к малым изменениям входа (через Якобиан).

**Плюс:** локально устойчивые представления.
**Минус:** тяжелее вычислительно.

---

### 5. Variational Autoencoder (VAE)
Это уже вероятностная версия:
- encoder выдает параметры распределения $q_\phi(z|x)$ (обычно $\mu, \sigma$)
- из распределения сэмплируют $z$
- decoder моделирует $p_\theta(x|z)$

Оптимизируют ELBO:

$$\mathcal{L}_{VAE} = \mathbb{E}_{q_\phi(z|x)}[\log p_\theta(x|z)] - D_{KL}(q_\phi(z|x)\|p(z))$$

**Плюс:** можно генерировать новые объекты.
**Минус:** сложнее обучение, риск "размытой" генерации.

---

## Вопрос 6: Чем автоэнкодер отличается от PCA?

- **PCA** — линейный метод
- **Autoencoder** — обычно нелинейный (через глубокие сети)

Если автоэнкодер линейный без нелинейностей и с MSE, он близок к PCA по смыслу. Но на сложных данных нелинейный encoder/decoder обычно сильнее.

---

## Вопрос 7: Как автоэнкодер используют для поиска аномалий?

Логика:
1. Обучаем на "нормальных" данных
2. Считаем reconstruction error на новых данных
3. Если ошибка выше порога, помечаем как аномалию

Скоринг:

$$s(x)=\|x-\hat{x}\|$$

Порог выбирают по валидации/квантили/целевому уровню false positives.

---

## Вопрос 8: Почему автоэнкодеры могут быть дорогими?

### 1. Два прохода архитектуры
Есть и encoder, и decoder. По сравнению с "обычным" классификатором часто больше вычислений на один объект.

### 2. Высокая размерность входа
Для изображений, аудио, длинных последовательностей reconstruction-задача очень ресурсоемкая.

### 3. Тяжелый decoder
В генеративных вариантах (особенно VAE с мощным декодером) стоимость инференса и обучения заметно растет.

### 4. Длинное и нестабильное обучение
Нужно балансировать качество реконструкции, регуляризацию, размер латента, чтобы избежать:
- переобучения
- деградации в identity mapping
- плохой генерации (для VAE)

---

## Вопрос 9: Типичные ошибки при работе с автоэнкодерами

- Слишком широкий bottleneck -> сеть копирует вход, признаки бесполезны
- Неверная нормализация данных -> плохая реконструкция
- Неподходящий loss (например, MSE там, где лучше BCE или перцептуальный лосс)
- Отсутствие контроля переобучения
- Неверно выбранный порог аномалий

---

## Вопрос 10: Мини-практика выбора архитектуры

- Табличные данные: MLP autoencoder
- Изображения: Convolutional autoencoder
- Последовательности/временные ряды: LSTM/GRU autoencoder или 1D-CNN autoencoder
- Генерация/латентные интерполяции: VAE

---

## Мини-резюме

1. Автоэнкодер учится реконструировать вход через сжатый латентный код
2. Главная ценность — полезное представление данных, а не просто "копия"
3. Для устойчивости и обобщения важны регуляризация и правильный bottleneck
4. Для аномалий ключевой сигнал — reconstruction error
5. Стоимость растет из-за encoder+decoder, размерности данных и сложности генеративных вариантов

- как они связаны с Transformer-архитектурой
- какие у них цели предобучения
- почему GPT чаще для генерации, а BERT/RoBERTa для понимания текста

# Блок: BERT и GPT — архитектура, обучение, применение

---

## Вопрос 1: Как BERT и GPT связаны с Transformer?

**Ответ:** обе модели построены на Transformer, но используют разные части архитектуры:

- **BERT**: в основном **encoder-only**
- **GPT**: в основном **decoder-only**

Это ключевое различие определяет:
- что модель видит в контексте
- как предобучается
- какие задачи решает лучше

---

## Вопрос 2: Что такое BERT и как он устроен?

**BERT** (Bidirectional Encoder Representations from Transformers) — модель понимания текста с двусторонним контекстом.

### Архитектура BERT
- Стек Transformer Encoder блоков
- Bidirectional self-attention: токен видит и левый, и правый контекст
- Специальные токены: `[CLS]`, `[SEP]`

Вход обычно формируется так:

`[CLS] sentence_A [SEP] sentence_B [SEP]`

где итоговый вектор `[CLS]` часто используют для классификации пары/текста.

---

## Вопрос 3: Как предобучается BERT?

Классически использовались две задачи:

### 1. Masked Language Modeling (MLM)
Случайно маскируют часть токенов и модель восстанавливает их по контексту.

$$\max \sum_{t \in M} \log p(x_t \mid x_{\setminus M})$$

где $M$ — множество замаскированных позиций.

### 2. Next Sentence Prediction (NSP)
Классификация: является ли предложение B продолжением A.

(В современных вариантах NSP часто заменяют/убирают.)

---

## Вопрос 4: Что умеет BERT лучше всего?

BERT силен в задачах **понимания**:
- классификация текста
- NER (разметка токенов)
- извлечение информации
- извлечение ответа в extractive QA
- semantic matching / reranking

Почему: bidirectional контекст дает очень богатые представления для каждого токена.

---

## Вопрос 5: Что такое GPT и как он устроен?

**GPT** (Generative Pretrained Transformer) — генеративная модель текста.

### Архитектура GPT
- Стек Transformer Decoder блоков
- Causal (masked) self-attention: токен на позиции $t$ видит только позиции $\le t$
- Авторегрессионная генерация: токен за токеном

Цель:

$$p(x) = \prod_{t=1}^{T} p(x_t \mid x_{<t})$$

То есть модель учится предсказывать следующий токен.

---

## Вопрос 6: Как предобучается GPT?

Основная задача — **next-token prediction**:

$$\max \sum_{t=1}^{T} \log p(x_t \mid x_{<t})$$

Это простая, но мощная objective:
- масштабируется на огромные корпуса
- напрямую готовит модель к генерации

---

## Вопрос 7: Что GPT делает лучше всего?

GPT особенно силен в:
- генерации текста
- диалогах
- суммаризации (instruction style)
- code generation
- творческих и multi-step задачах с длинным выводом

Почему: архитектура изначально заточена на причинную генерацию.

---

## Вопрос 8: BERT vs GPT — в чем ключевая разница?

### По направлению внимания
- **BERT**: двустороннее (лево+право)
- **GPT**: только в прошлое (causal)

### По objective
- **BERT**: masked reconstruction
- **GPT**: next-token prediction

### По типичным задачам
- **BERT**: понимание/классификация/разметка
- **GPT**: генерация/инструкции/диалог

---

## Вопрос 9: Почему BERT часто лучше в классификации, а GPT в генерации?

- Для классификации нужен максимально полный контекст фразы -> BERT видит обе стороны
- Для генерации нужна причинная последовательность -> GPT обучен предсказывать следующий токен

Именно цель предобучения создает "индуктивное смещение" под разные классы задач.

---

## Вопрос 10: Что дороже по вычислениям и где узкие места?

### На обучении
Обе модели дорогие. Стоимость растет по:
- числу слоев
- скрытой размерности
- длине последовательности
- объему данных

Self-attention в базовом виде по длине имеет стоимость порядка:

$$O(n^2 \cdot d)$$

### На инференсе
- **BERT**: часто один проход по входу (удобно для классификации)
- **GPT**: генерация токен за токеном, что увеличивает latency на длинных ответах

Для длинной генерации GPT может быть дороже именно из-за авторегрессии.

---

## Вопрос 11: Практический выбор модели

- Нужна **классификация, NER, matching** -> стартуй с BERT-like
- Нужна **генерация, чат, инструкт-модель** -> GPT-like
- Нужны обе стороны -> часто делают гибридный стек: encoder для retrieval/rerank + decoder для generation

---

## Вопрос 12: Короткая таблица сравнения

| Критерий | BERT | GPT |
|---|---|---|
| Архитектура | Encoder-only | Decoder-only |
| Attention | Bidirectional | Causal (left-to-right) |
| Предобучение | MLM (и исторически NSP) | Next-token prediction |
| Сильная сторона | Понимание текста | Генерация текста |
| Типичные задачи | Классификация, NER, QA-extractive | Диалог, генерация, code/text completion |
| Инференс | Обычно один прямой проход | Авторегрессия токен за токеном |

---

## Мини-резюме

1. BERT и GPT основаны на Transformer, но используют разные его части
2. BERT обучается понимать контекст (MLM), GPT — генерировать продолжение (causal LM)
3. BERT обычно лучше в discriminative NLP, GPT — в generative NLP
4. На практике выбор почти всегда определяется формулировкой задачи

# Продолжение: RoBERTa и современные GPT-подходы

---

## 1) Чем RoBERTa улучшает BERT

**RoBERTa** (Robustly Optimized BERT Pretraining Approach) сохраняет encoder-only идею BERT, но улучшает сам процесс предобучения.

### Ключевые отличия от оригинального BERT

- Убрали **NSP (Next Sentence Prediction)**
- Перешли к **dynamic masking** вместо фиксированного
- Обучали на существенно большем объеме данных
- Использовали большие batch size и более долгий тренинг
- Тщательнее настроили learning rate schedule

### Почему это обычно дает плюс к качеству

- MLM-сигнал становится богаче из-за dynamic masking
- Больше данных и шагов дают более устойчивые языковые представления
- Удаление NSP убирает не самый полезный pretraining objective

**Практический вывод:** если нужен сильный encoder-бейзлайн для классификации/NER/reranking, RoBERTa часто надежнее, чем "классический" BERT.

---

## 2) Современный пайплайн GPT: SFT -> RLHF -> Tool Use

### Шаг 1. SFT (Supervised Fine-Tuning)

Модель дообучают на парах "инструкция -> качественный ответ".

Что дает:
- лучшее следование формату и требованиям пользователя
- меньше случайного/хаотичного вывода
- более предсказуемый стиль ответа

### Шаг 2. RLHF (Reinforcement Learning from Human Feedback)

Схема:
1. Люди ранжируют несколько ответов модели
2. Обучают reward model
3. Основную модель оптимизируют так, чтобы максимизировать reward

Что дает:
- лучшее соответствие человеческим ожиданиям
- меньше токсичных/нежелательных ответов
- лучшее "поведение" в диалогах

### Шаг 3. Tool Use (вызов инструментов)

Модель не только генерирует текст, но и вызывает внешние инструменты:
- поиск
- калькулятор
- код-интерпретатор
- БД/CRM/API

Что это решает:
- повышает точность на фактических вопросах
- позволяет работать с актуальными данными
- снижает галлюцинации в операционных задачах

---

## 3) Как выбирать между BERT-like и GPT-like под продукт

Ниже практический чеклист в порядке принятия решения.

### Вопрос A: Нужна генерация длинного текста?
- Да -> GPT-like
- Нет, нужна классификация/извлечение/ранжирование -> BERT/RoBERTa-like

### Вопрос B: Какой latency budget?
- Жесткий realtime на CPU -> компактный encoder или небольшой distilled GPT
- Допустим выше latency ради качества -> более крупная модель, квантование/батчинг

### Вопрос C: Какой бюджет на инференс?
- Ограниченный бюджет -> меньшая модель + кэширование + маршрутизация запросов
- Бюджет позволяет -> гибрид (retrieval + генерация) или более крупная instruction model

### Вопрос D: Нужна ли объяснимость и стабильность предсказаний?
- Часто да в enterprise/регуляторике -> encoder-модели для классификации проще валидировать
- Для ассистентов/чатов важнее удобство диалога -> GPT-like

---

## 4) Быстрые продуктовые сценарии

- **Фильтрация тикетов/спама:** RoBERTa + linear head
- **NER и извлечение сущностей:** RoBERTa token-classification
- **Чат-ассистент:** GPT-like + SFT/RLHF + tool use
- **Поиск по базе знаний:** bi-encoder (retrieval) + cross-encoder rerank, при необходимости GPT для ответа
- **RAG в компании:** retrieval на encoder-эмбеддингах + GPT-like для синтеза ответа

---

## 5) Типичные ошибки выбора модели

- Берут большой GPT там, где достаточно дешевого encoder-классификатора
- Пытаются решать фактологию только генерацией без retrieval/tool use
- Не считают total cost of ownership: GPU, latency, поддержка, мониторинг
- Не делают offline/online A/B до раскатки

---

## Финальное мини-резюме

1. RoBERTa — это "усиленный BERT" за счет более грамотного pretraining
2. Современный GPT в проде обычно = SFT + RLHF + Tool Use
3. Выбор модели определяется задачей, latency и стоимостью, а не только "кто SOTA"
4. Для большинства реальных систем лучший результат дает не одна модель, а связка retrieval + encoder/reranker + generator
