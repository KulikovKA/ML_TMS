# ***Выводы о работе моделей***
> в тесте участвовало 4 видео:
>- видео с хорошим светом
>- видео с плохим светом
>- видео с синим светом(есть человек не из базы)
>- видео с красным светом( есть человек не из базы)

>на каждом видео 4 человека,чьи метки занесены в базу

## ***Face recognition***
1) **Видео с плохим светом**
- модель распознает сразу несколько лиц в кадре
- однако при движении лиц модель теряется
- при сильных гримасах и "кривляниях" модель не распознает лицо из базы и идентефицирует как "unknown"
- в некоторых кадрах модель также теряется и пишет "unknown"
- **Плохое освещение негативно сказывается на работе модели**
2) **Видео с хорошим светом**
- модель лучше распознает несколько лиц в кадре
- пропадет только при гримасах и то не при всех
- переодически при гримасах индентифицировала людей из базы как "unknown"
3) **Видео с синим светом**
- Синий свет сильно влияет на работу
- Модель часто теряется
- достаточно часто определается людей из базы как "unknown"
- модель смогла определить человека не из базы и записать его как "unknown"
4) **Видео с красным светом**
- Модель очень часто теряется
- Модель опредилила человека не из базы как человека из базы
- На кадре было две метки "egor"

## ***Haar cascade***
> модель не привязана к базе лиц
1) **Видео с плохим светом**
- Модель часто путает глаза с носом
- Плохо распознает лица
2) **Видео с хорошим светом**
- Модель работает лучше,но также путает глаза с носом и ртом
- Модель перепутала гербы на футболке с лицами
3) **Видео с красным светом**
- Очень много ошибок .Модель в хаосе.
