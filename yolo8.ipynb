# ***SDD VGG16 TEST №1***

```python
source_video_path = "C:\\Users\\kiril\\OneDrive\\Рабочий стол\\YOLO8_data\\457746_Rain_Rainy_1920x1080.mp4"
project_path = "C:\\Users\\kiril\\OneDrive\\Рабочий стол\\jupyter_projects\\realtime_models_runs\\SDD"
experiment_name = 'video_output1'
confidence_threshold = 0.5
output_path = os.path.join(project_path, experiment_name)
os.makedirs(output_path, exist_ok=True)
output_video_path = os.path.join(output_path, "result.mp4")
```

```python
cap = cv2.VideoCapture(source_video_path)

frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS))
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_video_path, fourcc, fps, (frame_width, frame_height))

print("Метка 1")
while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    img_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    transform = torchvision.transforms.ToTensor()
    input_tensor = transform(img_rgb).unsqueeze(0)

    with torch.no_grad():
        prediction = model(input_tensor)

    for i in range(len(prediction[0]['scores'])):
        score = prediction[0]['scores'][i]
        label_index = prediction[0]['labels'][i]
        
        if score > confidence_threshold and COCO_INSTANCE_CATEGORY_NAMES[label_index] == 'person':
            box = prediction[0]['boxes'][i].cpu().numpy().astype(int)
            cv2.rectangle(frame, (box[0], box[1]), (box[2], box[3]), (0, 255, 0), 2)
            
            score_text = f"person: {score:.2f}"
            cv2.putText(frame, score_text, (box[0], box[1] - 10), cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    out.write(frame)

cap.release()
out.release()
cv2.destroyAllWindows()
```

# Модель может работать в реальном времени,но требует большего кол-ва ресурсов.
Я оставил лишь демонстрация для одного видео.

```python

source_video_path = r"C:\Users\kiril\OneDrive\Рабочий стол\YOLO8_data\457746_Rain_Rainy_1920x1080.mp4"
confidence_threshold = 0.5

device = torch.device('cpu')
weights = SSD300_VGG16_Weights.DEFAULT
model = torchvision.models.detection.ssd300_vgg16(weights=weights).to(device).eval()
preprocess = weights.transforms()
categories = weights.meta["categories"]

win = "SSD300"
cv2.namedWindow(win, cv2.WINDOW_NORMAL)
cv2.resizeWindow(win, 960, 540)

cap = cv2.VideoCapture(source_video_path)

while cap.isOpened():
    ok, frame = cap.read()
    if not ok:
        break

    img_rgb = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    pil_img = Image.fromarray(img_rgb)
    tensor = preprocess(pil_img).to(device)

    with torch.no_grad():
        pred = model([tensor])[0]

    boxes = pred["boxes"].detach().cpu().numpy().astype(int)
    labels = pred["labels"].detach().cpu().numpy()
    scores = pred["scores"].detach().cpu().numpy()

    keep = scores >= confidence_threshold
    for box, lab, sc in zip(boxes[keep], labels[keep], scores[keep]):
        cls_name = categories[int(lab)]
        cv2.rectangle(frame, (box[0], box[1]), (box[2], box[3]), (0, 255, 0), 2)
        cv2.putText(frame, f"{cls_name}: {sc:.2f}", (box[0], max(0, box[1] - 10)),
                    cv2.FONT_HERSHEY_SIMPLEX, 0.5, (0, 255, 0), 2)

    disp = cv2.resize(frame, (960, 540), interpolation=cv2.INTER_AREA)
    cv2.imshow(win, disp)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

cap.release()
cv2.destroyAllWindows()
```

###  YOLO8
- быстрая обработка видео
- показывает высокий conf
- распознает всех людей,однако человека в далеко распознает не сразу,а по мере приближения кадра к нему 
### SDD VGG 16
- более долгая обработка видео
- распознает не все обьекты,даже те,которые хорошо видно,низкйи показатель conf.
- модель провалила тест

### YOLO 11l
- результаты схожи с yolo8 
-  чуть выше conf
- "проблемный" человек сразу обводиться в bb,однако переодически bb исчезает

### Итоговые выводы
- Лучшую полноту и стартовую уверенность по дальнему объекту показал YOLO11L, хотя стабильность BBox всё ещё не идеальна.  
- YOLOv8 остаётся быстрым и надёжным, но уступает YOLO11L по раннему захвату далёких целей.  
- SSD VGG16 в этом тесте не прошёл по скорости и полноте детекции, что делает его непрактичным для подобной сцены.
